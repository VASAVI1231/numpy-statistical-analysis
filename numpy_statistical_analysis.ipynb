```python
import numpy as np

# ---------------------------------------------------------
# NUMPY STATISTICAL ANALYSIS
# Internship Task 15 - AI & ML
# ---------------------------------------------------------

print("=" * 60)
print("NUMPY STATISTICAL ANALYSIS - STUDENT PERFORMANCE")
print("=" * 60)


# ---------------------------------------------------------
# 1. Create Dataset using NumPy
# ---------------------------------------------------------

student_ids = np.arange(1, 13)

math = np.array([78, 65, 92, 56, 88, 74, 95, 61, 83, 70, 54, 90])

science = np.array([82, 71, 89, 60, 91, 76, 94, 64, 85, 73, 58, 93])

english = np.array([75, 68, 86, 62, 84, 79, 90, 67, 81, 72, 59, 88])


# Combine all subject scores
scores = np.column_stack((math, science, english))


# ---------------------------------------------------------
# 2. Subject-wise Statistical Analysis
# ---------------------------------------------------------

subjects = ["Math", "Science", "English"]

for i, subject in enumerate(subjects):

    values = scores[:, i]

    print(f"\n--- {subject} ---")

    print(f"Mean              : {np.mean(values):.2f}")
    print(f"Median            : {np.median(values):.2f}")
    print(f"Standard Deviation : {np.std(values):.2f}")
    print(f"Minimum            : {np.min(values):.2f}")
    print(f"Maximum            : {np.max(values):.2f}")
    print(f"Range              : {np.ptp(values):.2f}")


# ---------------------------------------------------------
# 3. Overall Statistics
# ---------------------------------------------------------

all_scores = scores.flatten()

print("\n--- Overall Statistics ---")

print(f"Overall Mean       : {np.mean(all_scores):.2f}")
print(f"Overall Median     : {np.median(all_scores):.2f}")
print(f"Overall Std Dev    : {np.std(all_scores):.2f}")
print(f"Overall Minimum    : {np.min(all_scores):.2f}")
print(f"Overall Maximum    : {np.max(all_scores):.2f}")


# ---------------------------------------------------------
# 4. Student-wise Total and Average
# ---------------------------------------------------------

student_total = np.sum(scores, axis=1)

student_average = np.mean(scores, axis=1)


print("\n--- Student Performance ---")

for i in range(len(student_ids)):

    print(
        f"Student {student_ids[i]:2d} "
        f"| Total: {student_total[i]:3.0f} "
        f"| Average: {student_average[i]:.2f}"
    )


# ---------------------------------------------------------
# 5. Highest and Lowest Performing Students
# ---------------------------------------------------------

highest_index = np.argmax(student_average)

lowest_index = np.argmin(student_average)

print("\n--- Performance Comparison ---")

print(
    f"Highest Performing Student : "
    f"Student {student_ids[highest_index]} "
    f"({student_average[highest_index]:.2f})"
)

print(
    f"Lowest Performing Student  : "
    f"Student {student_ids[lowest_index]} "
    f"({student_average[lowest_index]:.2f})"
)


# ---------------------------------------------------------
# 6. Subject Comparison
# ---------------------------------------------------------

subject_means = np.mean(scores, axis=0)

best_subject_index = np.argmax(subject_means)

lowest_subject_index = np.argmin(subject_means)

print("\n--- Subject Comparison ---")

for i in range(len(subjects)):

    print(
        f"{subjects[i]:10s} "
        f"Average: {subject_means[i]:.2f}"
    )


print(
    f"\nBest Subject   : "
    f"{subjects[best_subject_index]}"
)

print(
    f"Lowest Subject : "
    f"{subjects[lowest_subject_index]}"
)


# ---------------------------------------------------------
# 7. Unusual Value Detection
# ---------------------------------------------------------

overall_mean = np.mean(all_scores)

overall_std = np.std(all_scores)

lower_limit = overall_mean - (2 * overall_std)

upper_limit = overall_mean + (2 * overall_std)


unusual_values = all_scores[
    (all_scores < lower_limit) |
    (all_scores > upper_limit)
]


print("\n--- Unusual Values Check ---")

print(f"Lower Limit : {lower_limit:.2f}")

print(f"Upper Limit : {upper_limit:.2f}")


if len(unusual_values) > 0:

    print(
        "Potential Unusual Values:",
        unusual_values
    )

else:

    print("No unusual values found.")


# ---------------------------------------------------------
# 8. Final Summary
# ---------------------------------------------------------

print("\n" + "=" * 60)

print("FINAL SUMMARY")

print("=" * 60)

print(f"Number of Students : {len(student_ids)}")

print(f"Overall Mean       : {np.mean(all_scores):.2f}")

print(f"Overall Median     : {np.median(all_scores):.2f}")

print(f"Overall Std Dev    : {np.std(all_scores):.2f}")

print(
    f"Best Student       : "
    f"Student {student_ids[highest_index]}"
)

print(
    f"Best Subject       : "
    f"{subjects[best_subject_index]}"
)

print("\nAnalysis completed successfully!")
```
